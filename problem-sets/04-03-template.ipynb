## 1. Data preparation and specified features

Run the supplied cell without changing it. It loads the data, gives the
variables readable names, rescales the monetary variables, groups legacy

Use all nine of these prepared features in the model:

- Categorical: `education_group` and `marriage_group`

The target is `default_next_month`. The client identifier is not a

```python
# RUN THIS CELL AS IS

PUBLIC_REPOSITORY = "okuchap/GB656_2026_public"
PUBLIC_REVISION = "main"


def course_data_source(file_name):
    """Return a local course-data path when available, otherwise its public URL."""
    for root in (Path.cwd(), *Path.cwd().parents):
        local_path = root / "data" / file_name
        if local_path.is_file():
            return local_path

    encoded_name = quote(file_name)
    return (
        "https://raw.githubusercontent.com/"
        f"{PUBLIC_REPOSITORY}/{PUBLIC_REVISION}/data/{encoded_name}"
    )


data_source = course_data_source("UCI_Credit_Card.csv")
credit = pd.read_csv(data_source).rename(
    columns={
        "LIMIT_BAL": "credit_limit",
        "EDUCATION": "education",
        "MARRIAGE": "marriage",
        "AGE": "age",
        "PAY_0": "pay_status_sep",
        "PAY_2": "pay_status_aug",
        "PAY_3": "pay_status_jul",
        "BILL_AMT1": "bill_amt_sep",
        "PAY_AMT1": "pay_amt_sep",
        "default.payment.next.month": "default_next_month",
    }
)

source_location = (
    "local course repository" if isinstance(data_source, Path) else "public GitHub repository"
)
print(
    f"Loaded {credit.shape[0]:,} rows and {credit.shape[1]} columns "
    f"from the {source_location}."
)

credit["credit_limit_10000"] = credit["credit_limit"] / 10_000
credit["bill_amt_sep_1000"] = credit["bill_amt_sep"] / 1_000
credit["pay_amt_sep_1000"] = credit["pay_amt_sep"] / 1_000

credit["education_group"] = pd.Categorical(
    credit["education"].map(
        {
            1: "graduate school",
            2: "university",
            3: "high school",
            4: "other",
            0: "other",
            5: "other",
            6: "other",
        }
    ).fillna("other"),
    categories=["graduate school", "high school", "other", "university"],
)
credit["marriage_group"] = pd.Categorical(
    credit["marriage"].map(
        {1: "married", 2: "single", 3: "other", 0: "other"}
    ).fillna("other"),
    categories=["married", "other", "single"],
)

print("Number of clients by outcome:")
display(credit["default_next_month"].value_counts().rename({0: "No default", 1: "Default"}))
print("\nShare of clients by outcome:")
display(
    credit["default_next_month"]
    .value_counts(normalize=True)
    .rename({0: "No default", 1: "Default"})
    .round(3)
)

credit[[
    "credit_limit_10000",
    "age",
    "pay_status_sep",
    "pay_status_aug",
    "pay_status_jul",
    "bill_amt_sep_1000",
    "pay_amt_sep_1000",
    "education_group",
    "marriage_group",
    "default_next_month",
]].head()
```
